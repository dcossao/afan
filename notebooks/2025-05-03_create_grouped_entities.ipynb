## Create dataset grouped_entities

Contais columns ID, role and entity. Entity is a string with semicolon-separated entities aggregated per ID and role.

For example, say roles_df has the following rows:

| index | ID  | role   | entity       |
|--------|-----|--------|-------------|
| 0      | 100 | victim | the planet  |
| 1      | 100 | victim | humanity    |


Then grouped_entities will have the next row:


| index | ID  | role   | entities              |
|-------|-----|--------|---------------------  |
| 0     | 100 | victim | the planet; humanity  |



```python
import pandas as pd
```

```python
roles_df = pd.read_csv(
    "../data/interim/roles_df.tsv", 
    sep='\t',
    encoding='ascii',
    encoding_errors='ignore'  # Skip non-ASCII characters
    )
```

No entity contains a semicolon

```python
roles_df[roles_df.entity.str.contains(";")]
```

Per ID and narrative role, concatenate entities separated by a semicolon

```python
# list didn't work, saved really funny
# grouped_entities = roles_df.groupby(['ID', 'role'])['entity'].agg(list).reset_index()

grouped_entities = roles_df.groupby(['ID', 'role'])['entity'].agg(lambda x: '; '.join(x)).reset_index()

grouped_entities.head()
```

Rename column entity to *entities*

```python
grouped_entities = grouped_entities.rename(columns={'entity': 'entities'})

grouped_entities.head()
```
