### Create train, dev and test sets per narrative role

```python
%load_ext autoreload
%autoreload 2
```

```python
from sklearn.model_selection import train_test_split

from afan.dataset import load_tsv
```

```python
news_with_roles = load_tsv(
    dataset_name="news_with_roles",
    data_dir="../data/interim"
)
```

### Split news_with_roles by narrative role

```python
def make_train_dev_test_sets(role: str) -> None:

    """
    - Split news_with_roles based on role (hero, villain, victim). 
    - Split that data into train, dev and test sets.
    - Save the train and dev sets to the processed data directory.
    - Save the test set to the test data directory.

    Parameters:
    ----------
    role : str
        The role to filter by ('hero', 'villain', or 'victim')

    Returns:
    -------
    tuple of (DataFrame, DataFrame, DataFrame)
        Returns train, dev and test sets split from the role-filtered data
    """

    role_df = news_with_roles[news_with_roles.role == role]

    # 25 percent of the data is set aside for testing
    train_dev, test = train_test_split(role_df, test_size=0.25, random_state=42, stratify=role_df["mbfc"])

    print(f"Number of articles in {role} test: {len(test)}")

    # 15 percent of the data will be used for prompt optimization. This set is called "train"
    train, dev = train_test_split(train_dev, test_size=0.85, random_state=42, stratify=train_dev["mbfc"])
    print(f"Number of articles in {role} train: {len(train)}")
    print(f"Number of articles in {role} dev: {len(dev)}")

    # save train and dev to processed data dir
    train.to_csv(
        f"../data/processed/{role}_train.tsv", 
        sep='\t', 
        index=False
    )
    dev.to_csv(
        f"../data/processed/{role}_dev.tsv", 
        sep='\t', 
        index=False
    )

    # save test to test data dir
    test.to_csv(f"../data/test/{role}_test.tsv", sep='\t', index=False)


```

### Heroes

```python
make_train_dev_test_sets("hero")
```

### Villains

```python
make_train_dev_test_sets("villain")
```

#### Victims

```python
make_train_dev_test_sets("victim")
```
